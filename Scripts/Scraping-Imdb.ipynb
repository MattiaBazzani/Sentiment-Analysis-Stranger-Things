```python
import csv
import time
import pandas as pd
from selenium import webdriver
from selenium.webdriver.support import expected_conditions as EC
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.common.action_chains import ActionChains
import warnings
warnings.filterwarnings('ignore')
```

```python
def extract_review_data(review_elements, i):
    review_data = []
    
    for x in review_elements:
        data = {}

        # Estrai il titolo
        titolo = x.find_elements(By.CLASS_NAME,"ipc-title__text")
        data["titolo"] = titolo[0].text

        # Estrai rating
        rt = x.find_elements(By.CLASS_NAME,"ipc-rating-star--rating")
        if rt:  # Check if the list has at least one element
            data["rating"] = rt[0].text
        else:
            data["rating"] = None

        # Estrai recensione
        recensione = x.find_elements(By.CLASS_NAME,"ipc-html-content-inner-div")
        data["recensione"]=recensione[0].text

        # estrai numero upvotes
        upvote = x.find_elements(By.CLASS_NAME,"ipc-voting__label__count--up")
        data["upvote"]=upvote[0].text

        # estrai numero downvotes
        downvote = x.find_elements(By.CLASS_NAME,"ipc-voting__label__count--down")
        data["downvote"]=downvote[0].text

        # inserire numero episodio
        data["episodio"]=i

        review_data.append(data)

    return review_data
```

```python
def data_frame_review_data(url, x, s=20) :

# 1.PERCOSO DEL DRIVER 
    service = webdriver.ChromeService('/Users/Hp_pc/OneDrive/Desktop/Uni/LM/1_2024-2025/SocialMedia/Lab/web_scraping/chromedriver.exe')
    driver = webdriver.Chrome(service=service)
    driver.get(url)

# 2.Clicca tasto accetta 
    time.sleep(10)
    driver.find_element(By.CSS_SELECTOR, '[data-testid="accept-button"]').click()

# 3.Click bottone per visualizzare tutti le  recensioni
    button = driver.find_element(By.XPATH, "//*[@id='__next']/main/div/section/div/section/div/div[1]/section[1]/div[3]/div/span[2]/button")

# Scorrere fino al pulsante
    driver.execute_script("arguments[0].scrollIntoView();", button)

# Attendi un breve periodo per permettere alla pagina di aggiornarsi
    time.sleep(4)

# Clicca sul pulsante
    button.click()

# 4. CERCA e CLICCA TUTTI i bottoni Spoiler

    time.sleep(s)
    print("\nCerca bottoni Spoiler...")
    buttons = driver.find_elements(By.TAG_NAME, "button")
    
    spoiler_count = 0
    spoiler_buttons_found = []
    
# Prima conta quanti bottoni Spoiler esistono
    for btn in buttons:
        try:
            btn_text = btn.text.strip().lower()
            if "spoiler" in btn_text:
                spoiler_buttons_found.append(btn)
        except:
            continue
    
    print(f"Trovati {len(spoiler_buttons_found)} bottoni Spoiler")
    
# Ora cliccali tutti
    for i, btn in enumerate(spoiler_buttons_found):
        try:
            # Fai scroll fino al bottone per essere visibile
            driver.execute_script("arguments[0].scrollIntoView({behavior: 'smooth', block: 'center'});", btn)
            time.sleep(3)
            
        # Clicca il bottone
            btn.click()
            spoiler_count += 1
            print(f"  ✓ Cliccato bottone Spoiler #{spoiler_count}")
            time.sleep(1)
            
        except Exception as e:
            print(f"  ✗ Errore cliccando bottone {i+1}: {str(e)[:50]}...")
            continue
    
    print(f"\n✅ Totale Spoiler cliccati: {spoiler_count}")

# 5.Estrai dati recensioni
    review_cards = driver.find_elements(By.CLASS_NAME, "user-review-item")

    review_data = extract_review_data(review_cards, x)

    df = pd.DataFrame(review_data)
    driver.close()
    return df
```
